##Evaluate line integrals using Green's Theorem

##Aim

**Evaluate the line integrals of the vector field along a curve using green's thorem**

**Problem 1**

Evaluate the line integral of the vector field $F(x,y)=(x^2-y,x+y^2)$ along the positively oriented unit circle C defined by $x^2+y^2=1$ using Green's Theorem .

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

#define the variables
x,y=sp.symbols('x,y')

#define the vector field components
P=x**2-y
Q= x+y**2

#compute the partial derivatives
dQ_dx=sp.diff(Q,x)
dP_dy=sp.diff(P,y)

# applying green's theorem
integrand= dQ_dx - dP_dy

#define the region D:x^2+y^2=1
r,theta=sp.symbols('r,theta')
x_polar=r*sp.cos(theta)
y_polar=r*sp.sin(theta)

#change of variables
integrand_polar=integrand.subs({x:x_polar,y:y_polar}) *r

#define the limits for r and theta
r_limits=(r,0,1)
theta_limits=(theta,0,2*sp.pi)

#compute the double i9ntegral in polar coordinates
area_integral =sp.integrate(integrand_polar,r_limits,theta_limits)

#display
area_integral




```

##Visualization of the vector field and region D

```python
#define the vector field components


def P(x,y):
    return x**2-y
def Q(x,y):
    return x+y**2

#create grid of points
x_vals=np.linspace(-1.5,1.5,20)
y_vals=np.linspace(-1.5,1.5,20)
X,Y=np.meshgrid(x_vals,y_vals)

#compute the vector field components at each point
U=P(X,Y)
V=Q(X,Y)

#define the unit circle
theta=np.linspace(0,2*np.pi,100)
x_circle=np.cos(theta)
y_circle=np.sin(theta)

#plot the vector field
plt.figure(figsize=(8,8))
plt.quiver(X,Y,U,V,color='Blue',alpha=0.5)

#plot the unit circle
plt.plot(x_circle,y_circle,color='Red',linewidth=2,label='Unit Circle')

#shade the area inside the unit circle
plt.fill_between(x_circle,y_circle,color='Red',alpha=0.2)

#add labels and legend
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Vector Field and Unit Circle')
plt.axhline(0,color='black',linewidth=0.5)
plt.axvline(0,color='black',linewidth=0.5)
plt.grid(color='grey',linestyle='--',linewidth = 0.5)
plt.legend()
plt.axis('equal')
plt.show()



```
