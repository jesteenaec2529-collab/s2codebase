# **Evaluating the line integral**

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

#Define the parameter
t = sp.symbols('t')

#Define the vector field f(x,y,z) =(P,Q,R)
x,y,z = sp.symbols('x y z')
P=x*y
Q=y*z
R=z*x

#Define the parameterized curve r(t) =(x(t),y(t),z(t))
x_t = sp.sin(t)
y_t = sp.cos(t)
z_t = t

#Define the vector feild as a function of t along
F=sp.Matrix([P,Q,R])
r_t=sp.Matrix([x_t,y_t,z_t])
F_r_t=F.subs({x:x_t,y:y_t,z:z_t})

#Compute the derivative of r(t) with respect to t

dr_t =sp.diff(r_t,t)

#Compute the dot product F(r(t)) .dr(t)

integrand =F_r_t.dot(dr_t)

#Define the limit of the parameter t
a=0 #start
b=sp.pi #end


#Compute the line integral
line_integral = sp.integrate(integrand,(t,a,b))

#display
line_integral


```
