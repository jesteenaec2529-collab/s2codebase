# **Experiment No:8**

Solve the  homogenous  linear differential equation

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

#define the symbols
t =sp.symbols('t')
y =sp.Function('y')(t)

#define the differential equation
diff_eq = sp.Eq(y.diff(t,t)+4*y.diff(t)+4*y,0)

#solve the differential equation
sol = sp.dsolve(diff_eq,y)

print("General solution:",sol)



```

```python
sol
```

```python
#apply initial condition
c1 ,c2 =sp.symbols('C1 C2')
y_0=sol.rhs.subs({t:0})
eqn1 =sp.Eq(y_0,1)
y_prime=sol.rhs.diff(t)
y_prime_0=y_prime.subs({t:0})
eqn2=sp.Eq(y_prime_0,-2)
solution=sp.solve((eqn1,eqn2),(c1,c2))
particular_sol=sol.rhs.subs(solution)
particular_solution=sp.Eq(y,particular_sol)
particular_solution
```

```python
import sympy as sp

# convert sympy solution to a numerical function
f=sp.lambdify(t,particular_sol,'numpy')

#time range for plotting
t_vals = np.linspace(0,10,400)

#calculate nemerical solution
y_vals=f(t_vals)

#plot the solution
plt.plot(t_vals,y_vals,label='y(t)')

plt.xlabel('t')
plt.ylabel('y(t)')
plt.title('Solution of y\'\' +4y\'+ 4y = 0')
plt.legend()
plt.grid(True)
plt.show()
```

## **$$m\dfrac{d^2x}{dt^2}+c \dfrac{dx}{dt}+kx=0$$**

## where $x(t)$ is the displacement of the mass as a function of time $t$ . **[Given $m=1 ,c=0.5,k=2$]**

solve the differential equation for the given initial  
$x(0)=1 $ and $ x'(0)= $

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

#define the symbols
t =sp.symbols('t')
x =sp.Function('x')(t)

#Given
m=1 ;c=0.5;k=2

#define the differential equation
diff_eq = sp.Eq(m*x.diff(t,t)+c*x.diff(t)+k*x,0)

#solve the differential equation
sol = sp.dsolve(diff_eq,x)

print("General solution:",sol)



```

```python
sol
```

```python
#apply initial condition
c1 ,c2 =sp.symbols('C1 C2')
x_0=sol.rhs.subs({t:0})
eqn1 =sp.Eq(x_0,1)
x_prime=sol.rhs.diff(t)
x_prime_0=x_prime.subs({t:0})
eqn2=sp.Eq(x_prime_0,0)
solution=sp.solve((eqn1,eqn2),(c1,c2))
particular_sol=sol.rhs.subs(solution)
particular_solution=sp.Eq(x,particular_sol)
particular_solution
```

```python
import sympy as sp

# convert sympy solution to a numerical function
f=sp.lambdify(t,particular_sol,'numpy')

#time range for plotting
t_vals = np.linspace(0,10,400)

#calculate nemerical solution
x_vals=f(t_vals)

#plot the solution
plt.plot(t_vals,x_vals,label='x(t)')

plt.xlabel('t')
plt.ylabel('x(t)')
plt.title('Solution of x\'\' +0.5x\'+ 2x = 0')
plt.legend()
plt.grid(True)
plt.show()
```
